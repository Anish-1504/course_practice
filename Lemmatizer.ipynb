### Wordnet Lemmatizer

```python
import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
"""
POS - Noun-n
      verb-v
      adjective-a
      adverb-r
"""

lemmatizer.lemmatize("going") #--> Noun-n
```

```python
lemmatizer.lemmatize("going", pos='v')
```

```python
lemmatizer.lemmatize("going", pos='a')
```

```python
lemmatizer.lemmatize("going", pos='r')
```

```python
words = ['eating', 'eats', 'eaten', 'writing', 'writes', 'programming', 'programs', 'finally', 'finalize']
```

```python
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word))
```

```python
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word, pos='v'))
```

```python
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word, pos='a'))
```

```python
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word, pos='r'))
```
