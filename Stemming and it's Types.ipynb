### Stemming

```python
words = ['eating', 'eats', 'eaten', 'writing', 'writes', 'programming', 'programs', 'finally', 'finalize']
```

### Porter Stemmer

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
for word in words:
    print(word+"---->"+stemming.stem(word))
```

```python
stemming.stem('congratulations') #---> Works very good for most of the words. But some words will not displayed correctly like below mentioned word. (Major disadv of Stemming)
```

### RegexStemmer Class

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer = RegexpStemmer('ing$|s$|e|$able', min=4 )
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem("ablenotable")
```

```python
reg_stemmer.stem('eatinge')
```

```python
from nltk.stem import SnowballStemmer
```

```python
snowballstemmer = SnowballStemmer('english')
```

```python
for word in words:
    print(word+"---->"+snowballstemmer.stem(word))
```

```python
stemming.stem("fairly"), stemming.stem("sportingly")
```
